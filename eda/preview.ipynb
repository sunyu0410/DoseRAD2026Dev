```python
import SimpleITK as sitk
import json
import matplotlib.pyplot as plt
```

```python
def parse_patient(pt_name):
    img = sitk.ReadImage(data_dir / pt_name / 'image' / 'ct.mha')
    ct = sitk.GetArrayFromImage(sitk.ReadImage(data_dir / pt_name / 'image' / 'ct.mha'))
    dose = sitk.GetArrayFromImage(sitk.ReadImage(data_dir / pt_name / 'dose' / 'Dose_B0_CP090.mha'))
    info = json.load(open(data_dir/f'{pt_name}/{pt_name}.json'))
    insocentre = info['beams'][0]['iso_center']
    i, j, k = isocentre_ijk = img.TransformPhysicalPointToIndex(insocentre)

    return ct, dose, k
    
```

```python
for pt_name in pts:
    ct, dose, k = parse_patient(pt_name)
    plt.imshow(ct[k], cmap='gray', alpha=0.5)
    plt.imshow(dose[k], alpha=0.5)
    plt.title(pt_name)
    plt.axis('off')
    plt.savefig(save_dir / f'{pt_name}.png')
    plt.clf()
```
